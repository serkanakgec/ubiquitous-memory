```python
# ============================================
# 🔥 AUTO OLLAMA CLOUD (PINGGY + WEB UI)
# ============================================

import subprocess, time, re

# --------------------------------------------
# 🧠 MODEL SEÇİN — aşağıdan birini seç:
# --------------------------------------------
#MODEL = "qwen2.5-coder:14b"      # büyük, güçlü, kod için
#MODEL = "gpt-oss:120b"           # orta boy
MODEL = "qwen3-coder:30b"   # hızlı, iyi
# MODEL = "llama3.1:8b"          # çok hızlı

print("🔧 Sistem hazırlanıyor...\n")

# --------------------------------------------
# 📌 PCI utils (bazı optimize paketler için)
# --------------------------------------------
subprocess.run("apt-get install -y pciutils", shell=True, stdout=subprocess.PIPE)

# --------------------------------------------
# 📌 Ollama kurulum
# --------------------------------------------
print("🧠 Ollama kuruluyor...")
subprocess.run("curl -fsSL https://ollama.com/install.sh | sh", shell=True)

print("🚀 Ollama başlatılıyor...")
subprocess.Popen(['ollama', 'serve'])
time.sleep(5)

# --------------------------------------------
# 📌 Model indir
# --------------------------------------------
print(f"📦 Model indiriliyor: {MODEL}")
subprocess.run(f"ollama pull {MODEL}", shell=True)

# --------------------------------------------
# 🌍 WebUI ve Pinggy Setup
# --------------------------------------------
print("\n🌐 Open WebUI yükleniyor...")
subprocess.run("pip install open-webui > /dev/null", shell=True)

# Install pinggy before importing it
print("\n🔗 Pinggy kuruluyor...")
subprocess.run("pip install pinggy", shell=True)

import pinggy # Moved import after installation

print("\n🚀 WebUI başlatılıyor...")
subprocess.Popen(["open-webui", "serve", "--port", "8000"])
time.sleep(5)

print("\n🔗 Pinggy tüneller oluşturuluyor...\n")
ollama_tunnel = pinggy.start_tunnel("localhost:11434")
webui_tunnel = pinggy.start_tunnel("localhost:8000")

# --------------------------------------------
# 🎉 RESULTS
# --------------------------------------------

print("====================================")
print("🎉 OLLAMA CLOUD MODE AKTİF")
print("====================================")
print(f"🧠 Model: {MODEL}")
print("\n🔗 API Endpoint:")
print(ollama_tunnel.urls)
print("\n🖥 Open WebUI (Tarayıcıdan aç):")
print(webui_tunnel.urls)
print("====================================")
print("\n⚠️ NOT: Bu hücreyi durdurma. Sunucu aktif kalır.")
```
